```python
import re
from cshogi.KIF import move_to_kif
def pv_to_kif(pv, sfen):
    moves_sequence = ""
    board = cshogi.Board(sfen)
    for i in range(len(pv)):
        if i == 0:
            ja_move = move_to_kif(board.move_from_usi(pv[i]), for_caption=True)
            board.push_usi(pv[i])
        else:
            prev_move = board.move_from_usi(pv[i-1])
            ja_move = move_to_kif(board.move_from_usi(pv[i]), prev_move, for_caption=True)
            board.push_usi(pv[i])
        
        if board.turn == cshogi.BLACK:
            turn = "▲" if i % 2 == 0 else "△"
        else:
            turn = "△" if i % 2 == 0 else "▲"
        formatted_move = turn + ja_move

        # ()とその中の数字を削除
        clean_move = re.sub(r"\(\d+\)", "", formatted_move)
        # "同 " を "同" に置換
        clean_move = re.sub(r"同\s+", "同", clean_move)

        # 「全」→「成銀」、「圭」→「成桂」、「杏」→「成香」に変換
        clean_move = clean_move.replace("全", "成銀").replace("圭", "成桂").replace("杏", "成香")

        moves_sequence += " " + clean_move
    return moves_sequence
```

```python
import cshogi
import re
from cshogi.KIF import move_to_kif
# --- dlshogiの呼び出し ---
def exec_dlshogi(input_sfen):
    engine = Engine("/workspace/DeepLearningShogi/usi/bin/usi")
    info_listener = InfoListener()
    engine.setoption("DNN_Model", "/workspace/model/model_resnet10_swish-072.onnx")
    engine.isready()
    engine.position(sfen=input_sfen)
    engine.go(byoyomi=1000, listener=info_listener)
    cp = info_listener.info["cp"]
    pv = info_listener.info["pv"]
    engine.quit()
    return {"cp":cp, "pv":pv}

# --- pvを棋譜表記に変換 ---
def pv_to_kif(pv, sfen):
    moves_sequence = ""
    board = cshogi.Board(sfen)
    root_turn = board.turn
    for i in range(len(pv)):
        if i == 0:
            ja_move = move_to_kif(board.move_from_usi(pv[i]), for_caption=True)
            board.push_usi(pv[i])
        else:
            prev_move = board.move_from_usi(pv[i-1])
            ja_move = move_to_kif(board.move_from_usi(pv[i]), prev_move, for_caption=True)
            board.push_usi(pv[i])
        
        if root_turn == cshogi.BLACK:
            turn = "▲" if i % 2 == 0 else "△"
        else:
            turn = "△" if i % 2 == 0 else "▲"
        formatted_move = turn + ja_move

        # ()とその中の数字を削除
        clean_move = re.sub(r"\(\d+\)", "", formatted_move)
        # "同 " を "同" に置換
        clean_move = re.sub(r"同\s+", "同", clean_move)

        # 「全」→「成銀」、「圭」→「成桂」、「杏」→「成香」に変換
        clean_move = clean_move.replace("全", "成銀").replace("圭", "成桂").replace("杏", "成香")

        moves_sequence += " " + clean_move
    return moves_sequence
```

```python
sfen = "ln1g2snk/1r4g1l/2pp1pbpp/p3psp2/1pP6/3PSP2P/PPB1P1PP1/2R1G1SK1/LN3G1NL b - 31"
engine_sfen = "sfen " + sfen
result_dlshogi = exec_dlshogi(engine_sfen)
```
